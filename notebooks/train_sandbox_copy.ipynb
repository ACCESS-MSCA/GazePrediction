```python
import numpy as np
import unisal.unisal
import torch
import os, json
from models import Conv_tsmixer
import cv2
from torchvision import transforms
import PIL
import re

# Convert the degrees to the range [0, 360]
def convert_hor_degree(degrees):
    return (degrees + 180 + 360) % 360


def convert_ver_degree(degrees):
    # -1 because the y-axis is inverted for opencv
    return ((-1) * degrees + 90 + 360) % 360


def calculate_angular_velocity(angles, sample_rate=30, vel_range=700):
    # Convert angles from degrees to radians
    angles_rad = np.deg2rad(angles)

    # Calculate the differences between consecutive angles
    angle_differences = np.diff(angles_rad)

    # Compute angular velocity in radians per second
    angular_velocity = angle_differences * sample_rate

    # Convert angular velocity to degrees per second
    angular_velocity = np.rad2deg(angular_velocity)

    # Clamp the angular velocity to the specified range
    angular_velocity = np.clip(angular_velocity, -vel_range, vel_range)

    # Add the first velocity again at the beginning to match the length of input angles
    angular_velocity = np.concatenate(([angular_velocity[0]], angular_velocity))

    return angular_velocity


def calculate_angle_diff(angles):
    angles_diff = np.diff(angles)
    angles_diff = np.concatenate(([0], angles_diff))
    return angles_diff

def preprocess_sal_img(img):
    img = np.ascontiguousarray(img[:, :, ::-1])

    preproc_cfg = {
        "out_size": [288, 384],
        "rgb_mean": (0.485, 0.456, 0.406),
        "rgb_std": (0.229, 0.224, 0.225),
    }
    transformations = [
        transforms.ToPILImage(),
        transforms.Resize(preproc_cfg["out_size"], interpolation=PIL.Image.LANCZOS),
        transforms.ToTensor(),
    ]
    if "rgb_mean" in preproc_cfg:
        transformations.append(
            transforms.Normalize(preproc_cfg["rgb_mean"], preproc_cfg["rgb_std"])
        )
    processing = transforms.Compose(transformations)
    tensor = processing(img)
    return tensor
```

```python
import cv2
import os
import json
import numpy as np
import pandas as pd
from glob import glob
import torch
from torch.utils.data import Dataset, DataLoader
import random

class GazePredDataset(Dataset):
    def __init__(
        self,
        video_paths,
        past_timepoints=15,
        future_timepoints=6,
        frame_sample_rate=5,
        sliding_window=1,
        original_fps=30,
    ):
        self.video_paths = video_paths
        self.past_timepoints = past_timepoints
        self.future_timepoints = future_timepoints
        self.sliding_window = sliding_window
        self.frame_sample_rate = frame_sample_rate
        self.original_fps = original_fps

    def __len__(self):
        return len(self.video_paths)

    def __getitem__(self, idx):
        video_path = self.video_paths[idx]
        frames = []
        head_positions = []
        prediction_sequences = []
        # load json metadata
        json_meta_path = os.path.splitext(video_path)[0] + ".json"
        if os.path.exists(video_path) and os.path.exists(json_meta_path):
            with open(json_meta_path, "r") as file:
                json_data = json.load(file)
            metadata = pd.DataFrame(json_data)

            # calculate velocities and angle diffs in all 5 sec videos
            angles_horz = metadata["horz_hmd_deg"].to_numpy()
            hmd_ang_vel_horz = calculate_angular_velocity(angles_horz)
            angles_vert = metadata["vert_hmd_deg"].to_numpy()
            hmd_ang_vel_vert = calculate_angular_velocity(angles_vert)
            hmd_ang_diff_horz = calculate_angle_diff(angles_horz)
            hmd_ang_diff_vert = calculate_angle_diff(angles_vert)
            giw_horz = metadata["giw_long"].to_numpy()
            giw_vert = metadata["giw_lat"].to_numpy()
            filename = metadata["cropped_frame_filename"].to_numpy()[0]
            # start frame extraction
            cap = cv2.VideoCapture(video_path)
            # fnmber is the present frame
            for fnumber in range(
                self.past_timepoints,
                len(metadata) - self.future_timepoints,
                self.sliding_window,
            ):
                images_seq = []
                # Gather frames
                # Present frame is taken last from the for loop
                for i in reversed(range(self.past_timepoints)):
                    cap.set(cv2.CAP_PROP_POS_FRAMES, fnumber - i)
                    ret, frame = cap.read()
                    if not ret:
                        print("Failed to grab frame")
                        break
                    frame = preprocess_sal_img(frame)
                    images_seq.append(frame)
                # Gather hmd potional metadata
                hmd_seq = list(
                    zip(
                        hmd_ang_vel_horz[fnumber + 1 - self.past_timepoints : fnumber + 1],
                        hmd_ang_vel_vert[fnumber + 1 - self.past_timepoints : fnumber + 1],
                        hmd_ang_diff_horz[fnumber + 1 - self.past_timepoints : fnumber + 1],
                        hmd_ang_diff_vert[fnumber + 1 - self.past_timepoints : fnumber + 1],
                    )
                )
                pred_seq = list(
                    zip(
                        (
                            giw_horz[fnumber + 1 : fnumber + self.future_timepoints + 1]
                            - angles_horz[fnumber]
                        )
                        / 360,
                        (
                            giw_vert[fnumber + 1 : fnumber + self.future_timepoints + 1]
                            - angles_vert[fnumber]
                        )
                        / 180,
                    )
                )
                # Gather all inputs
                frames.append(images_seq)
                head_positions.append(hmd_seq)
                prediction_sequences.append(pred_seq)
            cap.release()
            return (
                torch.tensor(np.array(frames), dtype=torch.float32),
                torch.tensor(np.array(head_positions), dtype=torch.float32),
                torch.tensor(np.array(prediction_sequences), dtype=torch.float32),
                filename
            )


def collate_fn(batch, minibatch=16):
    # Filter out None samples
    batch = [sample for sample in batch if sample is not None]
    if len(batch) == 0:
        return None  # Return None if all samples are None

    frames_batch, head_positions_batch, predictions, filename = zip(*batch)
    # Flatten the first dimension
    frames_flat = torch.cat(frames_batch, dim=0)
    head_positions_flat = torch.cat(head_positions_batch, dim=0)
    predictions_flat = torch.cat(predictions, dim=0)
    # Shuffle the flattened data
    indices = torch.randperm(frames_flat.size(0))
    frames_flat = frames_flat[indices]
    head_positions_flat = head_positions_flat[indices]
    predictions_flat = predictions_flat[indices]
    # Split batch into chunks of batch_size
    batches = [
        [
            frames_flat[i : i + minibatch],
            head_positions_flat[i : i + minibatch],
            predictions_flat[i : i + minibatch],
        ]
        for i in range(0, len(frames_flat), minibatch)
    ]
    # Filter out any smaller batches
    batches = [
        b for b in batches if len(b[0]) == minibatch
    ]
    return batches


# Function to create the dataloader
def get_dataloader(video_paths, batch_size=32, minibatch=32, 
                   past_timepoints=15, future_timepoints=6,
                   frame_sample_rate=5, sliding_window=5,
                   original_fps=30, shuffle=False, num_workers=8):
    dataset = GazePredDataset(video_paths,
                                past_timepoints=past_timepoints,
                                future_timepoints=future_timepoints,
                                sliding_window=sliding_window,
                                frame_sample_rate=frame_sample_rate,
                                original_fps=original_fps)
    dataloader = DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=shuffle,
        num_workers=num_workers,
        collate_fn=lambda batch: collate_fn(batch, minibatch),
    )
    return dataloader
```

```python
def load_unisal(sal_path, device):
    unisal_model = unisal.unisal.model.UNISAL()
    unisal_model.load_state_dict(torch.load(sal_path))
    unisal_model.to(device)
    unisal_model.eval()
    # Freeze the weights of the UNISAL model
    for param in unisal_model.parameters():
        param.requires_grad = False
    
    return unisal_model
unisal_model = load_unisal("./models/unisal.pth", 'cuda')
```

```python
import os
import json
import cv2
import torch
import numpy as np
from torch.utils.data import Dataset, DataLoader
from natsort import natsorted
from glob import glob


class CustomDataset(Dataset):
    def __init__(self, vd_paths, k=10, f=2):
        self.vd_paths = vd_paths
        self.k = k
        self.f = f

    def __len__(self):
        return len(self.vd_paths)

    def __getitem__(self, idx):
        vd_path = self.vd_paths[idx]
        frames = []
        head_positions = []

        frame_idx = self.k - 1
        json_meta_path = os.path.splitext(vd_path)[0] + ".json"

        if os.path.exists(vd_path) and os.path.exists(json_meta_path):
            with open(json_meta_path, "r") as file:
                json_data = json.load(file)

            cap = cv2.VideoCapture(vd_path)
            cap.set(cv2.CAP_PROP_POS_FRAMES, int(frame_idx))
            while cap.isOpened():
                b_frames = []
                for _ in range(self.f):
                    ret, frame = cap.read()
                    if not ret:
                        break
                    frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
                    b_frames.append(frame)
                    frame_idx += 1
                else:
                    head_positions.append(
                        [
                            [
                                data["frame_number"],
                                data["video_id"],
                                data["user_id"],
                                data["task_id"],
                                data["giw_long"],
                                data["giw_lat"],
                            ]
                            for data in json_data[
                                frame_idx - self.k - 1 : frame_idx - 1
                            ]
                        ]
                    )
                    frames.append(b_frames)
                    continue
                break
            else:
                raise ValueError("Failed to read frame from video:", vd_path)

            cap.release()

        return torch.tensor(frames), torch.tensor(head_positions)


def custom_collate(batch):
    # Filter out None samples
    batch = [sample for sample in batch if sample is not None]
    if len(batch) == 0:
        return None  # Return None if all samples are None

    frames_batch, head_positions_batch = zip(*batch)

    # Flatten the first dimension
    frames_flat = torch.cat(frames_batch, dim=0)
    head_positions_flat = torch.cat(head_positions_batch, dim=0)

    # # Shuffle the flattened data
    # indices = torch.randperm(frames_flat.size(0))
    # frames_flat = frames_flat[indices]
    # head_positions_flat = head_positions_flat[indices]

    # Define your batch sizes
    batch_sizes = [32]  # Example batch sizes

    # Create batches of different sizes
    batches = []
    start = 0
    for batch_size in batch_sizes:
        end = start + batch_size
        batches.append((frames_flat[start:end], head_positions_flat[start:end]))
        start = end

    return batches


# Define your vd_paths
vd_paths = natsorted(glob("../data/results/EHT_dataset_salvideos/**/**/*.mp4"))

# Create an instance of your custom dataset
dataset = CustomDataset(vd_paths)

# Create a PyTorch DataLoader
dataloader = DataLoader(
    dataset,
    batch_size=4,
    shuffle=True,
    pin_memory=True,
    num_workers=4,
    collate_fn=custom_collate,
    drop_last=True,
)


batch_num = 0
for batch in dataloader:
    # if batch is None:
    #     continue
    frames, head_positions = batch[0]
    batch_num += 1
    print(f"Batch {batch_num}")
    print(np.array(frames).shape, np.array(head_positions).shape)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.utils.tensorboard import SummaryWriter
from torchtsmixer import TSMixer

from torchinfo import summary

import sys
from natsort import natsorted
from glob import glob
sys.path.append('..')
from dataloader import get_dataloader
import numpy as np

%load_ext autoreload
%autoreload 2

import warnings
warnings.filterwarnings('ignore')
from tqdm import tqdm

class Conv_tsmixer(nn.Module):
    def __init__(
        self,
        num_time_features,
        conv_output_embeddings,
        sequence_length,
        prediction_length,
        prediction_features,
        image_kernel_size=5,  # Default kernel size for convolutions
        image_padding=0,
    ):
        super(Conv_tsmixer, self).__init__()

        # Convolutional layers for each image (separate branches)
        self.conv1 = nn.Conv2d(
            1, 16, kernel_size=image_kernel_size, padding=image_padding
        )
        self.bn1 = nn.BatchNorm2d(16)
        self.conv2 = nn.Conv2d(
            16,
            8,
            kernel_size=image_kernel_size,
            padding=image_padding,
        )
        self.bn2 = nn.BatchNorm2d(8)
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2, padding=0)

        self.fc1 = nn.Linear(8 * 3 * 3, conv_output_embeddings)

        # Input channel size adjusted to consider individual image features
        self.tsmixer = TSMixer(
            sequence_length=sequence_length,
            prediction_length=prediction_length,
            input_channels=conv_output_embeddings
            + num_time_features,  # Adjusted input channels
            output_channels=prediction_features,
        )

    def forward(self, image, time_series):
        # Image processing for each image (separate branches)
        batch_size = image.size(0)
        num_images = image.size(1)
        num_time_steps = time_series.size(1)

        image_features = []
        for i in range(num_images):  # Process each image in the batch
            img = image[:, i, :, :]  # Select the ith image
            img = img.unsqueeze(1)  # Add channel dimension
            x = F.interpolate(img, size=(25, 25))  # Resize
            x = self.pool(F.relu(self.bn1(self.conv1(x))))
            x = self.pool(F.relu(self.bn2(self.conv2(x))))
            x = x.view(batch_size, -1)  # Flatten
            x = F.relu(self.fc1(x))
            # Expand features to match time series length and concatenate with time series data
            image_features.append(x.unsqueeze(1))
        img_features = torch.cat(image_features, dim=1)
        # print(img_features.shape)

        linspace_imgs_to_time_series = torch.linspace(
            0, num_time_steps, num_images + 1, dtype=torch.int
        )
        features = torch.zeros(8, 15, 10, device="cuda")

        # Copy the slices from A to the appropriate ranges in B
        for i in range(num_images):
            start_idx = linspace_imgs_to_time_series[i]
            end_idx = linspace_imgs_to_time_series[i + 1]
            features[:, start_idx:end_idx, :] = (
                img_features[:, i, :].unsqueeze(1).repeat(1, end_idx - start_idx, 1)
            )

        # print(features.shape)  # Should print torch.Size([8, 15, 10])
        # Concatenate time series data with features from all images
        combined_features = torch.cat((time_series, features), dim=2)
        # print(combined_features.shape)
        # # Pass the concatenated features through the TSMixer
        output = self.tsmixer(combined_features)

        return output



# Example usage
if __name__ == "__main__":
    batch_size = 8
    num_workers = 4
    num_time_features = 4
    conv_output_embeddings = 10
    sequence_length = 15
    prediction_length = 1
    prediction_features = 4
    num_epochs = 10
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    # Create dummy inputs
    # image_input = torch.randn(batch_size, 3, 1, 224, 224)
    # time_series_input = torch.randn(batch_size, sequence_length, num_time_features)

    # Initialize the model
    model = Conv_tsmixer(
        num_time_features,
        conv_output_embeddings,
        sequence_length,
        prediction_length,
        prediction_features,
    ).to(device)

    # Loss function and optimizer
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=0.0001)

    # Define your vd_paths
    vd_paths = natsorted(glob(
            "/home/alien/Documents/access/processed_dataset/eht_dataset_salvideos/**/**/*.mp4"
        ))
    
    # Split the video paths into training and validation sets
    train_vd_paths = vd_paths[:int(0.8 * len(vd_paths))]
    val_vd_paths = vd_paths[int(0.8 * len(vd_paths)):]

    train_dataloader = get_dataloader(
        train_vd_paths, batch_size=batch_size, num_workers=num_workers
    )
    val_dataloader = get_dataloader(
        val_vd_paths, batch_size=batch_size, num_workers=num_workers
    )

    # TensorBoard writer
    writer = SummaryWriter()

    for epoch in tqdm(range(num_epochs)):
        model.train()
        epoch_loss = 0.0
        for batches in train_dataloader:
            for b in batches:
                images, timeseries, predictions = b
                images, timeseries, predictions = images.to(device), timeseries.to(device), predictions.to(device)
                # Zero the parameter gradients
                optimizer.zero_grad()

                # Forward pass
                outputs = model(images, timeseries)
                loss = criterion(outputs, predictions)

                # Backward pass and optimize
                loss.backward()
                optimizer.step()
                
                epoch_loss += loss.item()
        
        avg_epoch_loss = epoch_loss / len(train_dataloader)
        print(f'Epoch [{epoch+1}/{num_epochs}], Training Loss: {avg_epoch_loss:.4f}')
        
        # Log the training loss to TensorBoard
        writer.add_scalar('Training Loss', avg_epoch_loss, epoch)
        
        # Validation step
        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for batches in val_dataloader:
                for b in batches:
                    images, timeseries, predictions = b
                    images, timeseries, predictions = images.to(device), timeseries.to(device), predictions.to(device)
                    
                    # Forward pass
                    outputs = model(images, timeseries)
                    loss = criterion(outputs, predictions)
                    
                    val_loss += loss.item()
        
        avg_val_loss = val_loss / len(val_dataloader)
        print(f'Epoch [{epoch+1}/{num_epochs}], Validation Loss: {avg_val_loss:.4f}')
        
        # Log the validation loss to TensorBoard
        writer.add_scalar('Validation Loss', avg_val_loss, epoch)

    # Close the TensorBoard writer
    writer.close()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchtsmixer import TSMixer

# Model parameters
sequence_length = 10
prediction_length = 5
input_channels = 2
output_channels = 1

# Create the TSMixer model
model = TSMixer(sequence_length, prediction_length, input_channels, output_channels)

# Loss function and optimizer
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.0001)

# Dummy dataset (replace with real data)
# Assuming batch_size, seq_len, num_features format
X_train = torch.randn(10, 32, sequence_length, input_channels)
y_train = torch.randn(10, 32, prediction_length, output_channels)

# Training loop
num_epochs = 10
for epoch in range(num_epochs):
    model.train()
    for X, y in zip(X_train, y_train):
        # Zero the parameter gradients
        optimizer.zero_grad()

        # Forward pass
        outputs = model(X)
        loss = criterion(outputs, y)

        # Backward pass and optimize
        loss.backward()
        optimizer.step()

    print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}")

print("Training complete")
```
